#### 2. tupla: estatísticas

```python
numeros = (12, 5, 8, 20, 15, 3, 10)

menor = numeros[0]
maior = numeros[0]
soma = 0

for num in numeros:
    soma += num
    if num > maior:
        maior = num
    elif num < menor:
        menor = num

media = soma / len(numeros)

print("menor valor:", menor)
print("maior valor:", maior)
print("soma:", soma)
print("média:", media)
```

#### 4. conjunto: operações básicas

```python
A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7, 8}

uniao = A.union(B)
intersecao = A.intersection(B)
rest_A = A.difference(B)
rest_B = B.difference(A)
dif_simetrica = uniao.difference(intersecao)
print("união:", uniao)
print("interseção:", intersecao)
print("elementos pertencentes somente a A:", rest_A)
print("elementos pertencentes somente a B:", rest_B)
print("diferença simétrica:", dif_simetrica)
```

#### 6. conjunto: testando subconjuntos

```python
A = {1, 2, 3, 4, 5, 6}
B = {2, 4, 6} 
C = {1, 2, 7}

eh_subconjunto_A = True
for elemento in B:
    if elemento not in A:
        eh_subconjunto_A = False
        break
print("B é subconjunto de A:", eh_subconjunto_A)

eh_subconjunto_C = True
for elemento in C:
    if elemento not in A:
        eh_subconjunto_C = False
        break
print("C é subconjunto de A:", eh_subconjunto_C)
```

#### 7. dicionários: notas dos alunos

```python
notas_alunos = {
    "Leo": 8,
    "Tales": 9,
    "Letícia": 9,
    "Erick": 8,
    "Amanda": 7
}

# imprime todos os alunos e suas notas
for aluno, nota in notas_alunos.items():
    print(f"{aluno}: {nota}")

# calcula a média das notas
media = sum(notas_alunos.values()) / len(notas_alunos)
print(f"Média das notas: {media:.2f}")

# encontra o aluno com a maior nota
melhor_aluno = max(notas_alunos, key=notas_alunos.get)
print(f"Aluno com a maior nota: {melhor_aluno}")
```

#### 8. dicionário: contagem de palavras

```python
import re

frase = "python é fácil, e python é poderosa"
palavras = re.split(r'[,\s]+', frase)
frequencia = {}
for palavra in palavras:
    if palavra in frequencia:
        frequencia[palavra] += 1
    else:
        frequencia[palavra] = 1
for palavra, contagem in frequencia.items():
    print(f"{palavra}: {contagem}")
print(palavras)
print(frequencia)
```

#### 11. dicionário de conjuntos

```python
estudantes = {
    "Alice": {"Python", "Algoritmos", "Bancos de Dados"},
    "Roberto": {"Python", "Redes de Computadores"},
    "Carlos": {"Algoritmos", "Bancos de Dados"},
    "Diana": {"Python", "Algoritmos", "Redes de Computadores"}
}
print(estudantes.items())

# todos os alunos que cursam Python
alunos_python = {aluno for aluno, cursos in estudantes.items() if "Python" in cursos}
print("Alunos que cursam Python:", alunos_python)

# todos os alunos que cursam Algoritmos
alunos_algoritmos = {aluno for aluno, cursos in estudantes.items() if "Algoritmos" in cursos}
print("Alunos que cursam Algoritmos:", alunos_algoritmos)

# todas as disciplinas cursadas por todos os alunos
disciplinas = {disciplina for cursos in estudantes.values() for disciplina in cursos}
print("Disciplinas cursadas por todos os alunos:", disciplinas)
```

#### 13. análise de palavras

```python
import re

texto = "Em 12 de agosto de 2026, milhões de pessoas na Islândia e no norte da Espanha acompanharam um raro eclipse solar total, o primeiro visível na Europa em 27 anos. Na Espanha, o fenômeno atraiu milhões de visitantes, enquanto a Islândia recebeu dezenas de milhares de observadores. O eclipse durou mais de dois minutos em algumas regiões e transformou o evento astronômico em uma grande celebração científica e turística."

# encontra o numero de palavras diferentes
palavras = re.findall(r'\b\w+\b', texto.lower())
print("Número de palavras diferentes:", len(set(palavras)))

# frequencia de cada palavra
frequencia = {}
for palavra in palavras:
    frequencia[palavra] = frequencia.get(palavra, 0) + 1

print("Frequência de cada palavra:", frequencia)

# palavra mais frequente
palavra_mais_frequente = max(frequencia, key=frequencia.get)
print("Palavra mais frequente:", palavra_mais_frequente, "com", frequencia[palavra_mais_frequente], "ocorrências")

# conjunto de palavras que aparecem apenas uma vez
palavras_uma_vez = {palavra for palavra, contagem in frequencia.items() if contagem == 1}
print("Palavras que aparecem apenas uma vez:", palavras_uma_vez)
```

#### 15. sistema de estoque

```python
estoque = {
    "maçãs": 25,
    "bananas": 12,
    "laranjas": 18,
    "peras": 7
}

# adiciona 10 maçãs
estoque["maçãs"] += 10

# remove 5 bananas
estoque["bananas"] -= 5

# adiciona 20 uvas
estoque["uvas"] = 20

# imprime os produtos com menos de 10 unidades
for produto, quantidade in estoque.items():
    if quantidade < 10:
        print(f"{produto}: {quantidade}")

# calcula o número total de itens
total_itens = sum(estoque.values())
print("Número total de itens em estoque:", total_itens)
```

#### 16. sistema de matrículas: desafio

```python
disciplinas = {
    "Python": {"Alice", "Roberto", "Carlos"},
    "Algoritmos": {"Diana", "Carlos", "Roberto"},
    "Banco de Dados": {"Alice", "Diana", "Eva"},
    "Redes de Computadores": {"Roberto", "Eva"}
}

# encontra todos os alunos matriculados em pelo menos uma disciplina
todos_alunos = set()
for alunos in disciplinas.values():
    todos_alunos.update(alunos)
print("Alunos matriculados em pelo menos uma disciplina:", todos_alunos)

# encontra os alunos simultaneamente matriculados em python e algoritmos
alunos_python = disciplinas["Python"]
alunos_algoritmos = disciplinas["Algoritmos"]
alunos_python_e_algoritmos = alunos_python.intersection(alunos_algoritmos)
print("Alunos matriculados em Python e Algoritmos:", alunos_python_e_algoritmos)

# encontra os alunos matriculados em exatamente uma disciplina
alunos_uma_disciplina = set()
for aluno in todos_alunos:
    disciplinas_aluno = [disciplina for disciplina, alunos in disciplinas.items() if aluno in alunos]
    if len(disciplinas_aluno) == 1:
        alunos_uma_disciplina.add(aluno)

# encontra o número de disciplinas cursadas por cada aluno
for aluno in todos_alunos:
    disciplinas_aluno = [disciplina for disciplina, alunos in disciplinas.items() if aluno in alunos]
    print(f"Aluno {aluno} cursa {len(disciplinas_aluno)} disciplina(s): {', '.join(disciplinas_aluno)}")
```
