1. determinante de uma matriz

```python
import numpy as np 
from scipy import linalg as la

A = np.array([
    [4, 2],
    [3, 5]
])

print(la.det(A))
```

2. resolver um sistema linear

```python
A = np.array([
    [2, 1],
    [1, 3]
])

b = np.array([5, 6])

x = la.solve(A, b)

print(x)
```

3. autovalores

```python
A = np.array([
    [4, 1],
    [2, 3]
])

avalores, avetores = la.eig(A)

print(f'autovalores de A: {avalores}')
print(f'autovetores de A: {avetores}')
```

4. integração numérica

```python
from scipy.integrate import quad

f = lambda x: x**3

I, erro = quad(f, 0, 2)

print(I)
print(erro)
```

5. cálculo de raízes de funções não lineares

```python
from scipy.optimize import root_scalar

f = lambda x: x**2 - 7

resultado = root_scalar(f, bracket=[2, 3])

print(resultado)
```

6. minimização

```python
from scipy.optimize import minimize

f = lambda x: x**2 - 6*x + 11

resultado = minimize(f, x0=2)

print(f'o mínimo ocorre em x={resultado.x}')
print(f'o valor mínimo é {resultado.fun}')
```

7. interpolação

```python
from scipy.interpolate import interp1d 

pontos = [(0,0), (1,1), (2,4), (3,9)]
x, y = zip(*pontos)

f = interp1d(x, y)

print(f(1.5))
```

10. distância euclidiana

```python
from scipy.spatial.distance import euclidean

p = [1, 2, 3]
q = [4, 6, 3]

d = euclidean(p, q)

print(d)
```

11. criar uma matriz esparsa

```python
from scipy.sparse import coo_array

valores = [1, 2, 3, 4]
linhas = [0, 1, 3, 3]
colunas = [0, 2, 0, 3]

A = coo_array((valores, (linhas, colunas)), shape=(4, 4))
print(A)
```

12. sistema linear esparso

```python
from scipy.sparse import csr_array
from scipy.sparse.linalg import spsolve

A = csr_array([
    [4, 1, 0],
    [1, 4, 1],
    [0, 1, 4]
])

b = np.array([6, 6, 6])

x = spsolve(A, b)

print(x)
```

13. transformada de Fourier

```python
from scipy.fft import fft 

x = np.array([1, 1, 1, 1])

X = fft(x)

print(X)
```

14. resolver uma equação diferencial ordinária

```python
from scipy.integrate import solve_ivp 

def f(t, y):
    return -y

t_span = [0, 5]
y0 = [1]

sol = solve_ivp(
    f,
    t_span,
    y0
)

print(sol)
print(f'y(5) = {sol.y[0, -1]}')
print(f'erro: {np.abs(sol.y[0, -1] - np.exp(-5))}')
```

15. ajuste de uma curva

```python
from scipy.optimize import curve_fit

x = np.array([0, 1, 2, 3, 4])
y = np.array([1, 3, 5, 7, 9])

def r(x, a, b):
    return a*x + b

popt, pcov = curve_fit(r, x, y)

print(f'a = {popt[0]}\nb = {popt[1]}')
```

17. decomposição QR

```python
from scipy.linalg import qr, solve_triangular

A = np.array([
    [1, 1],
    [1, 0],
    [0, 1]
])

b = np.array([2, 3, 4])

Q, R = qr(A, mode='economic')
x = solve_triangular(R, Q.T @ b)

print(f'solução mínimos quadrados: {x}')
```

18. SVD

```python
from scipy.linalg import svd 

A = np.array([
    [3, 0],
    [0, 2]
])

U, S, VT = svd(A)

print(U @ np.diag(S) @ VT)
```
