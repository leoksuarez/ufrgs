1. função para calcular o dobro

```python
def dobro(n):
    return 2*n

dobro(7)
```

2. função para calcular a média

```python
def media(a, b, c):
    return (a+b+c)/3

media(3,2,7)
```

3. função para verificar paridade

```python
def eh_par(n):
    if n % 2 == 0:
        return "é par"
    else:
        return "não é par"
eh_par(7)
```

4. conversão de temperatura

```python
def celsius_para_fahrenheit(c):
    return (9/5)*c + 32

celsius_para_fahrenheit(30)
```

5. área de um retângulo

```python
def area_retangulo(base, altura):
    return base*altura

area_retangulo(4, 5)
```

6. parâmetro opcional

```python
def saudacao(nome, mensagem='olá,'):
    print(mensagem, nome)
saudacao('Rudnei')
```

7. potência com expoente padrão

```python
def potencia(base, expoente=2):
    return base**expoente

potencia(4, 3)
```

8. argumentos posicionais e nomeados

```python
def apresentar(nome, idade, cidade):
    print(f'Olá, meu nome é {nome}, tenho {idade} anos e moro em {cidade}.')

# a) usando apenas argumentos posicionais
apresentar('Leo', 23, 'Porto Alegre')

# b) usando apenas argumentos nomeados
apresentar(idade=79, cidade="São José dos Campos", nome="Laura")

# c) usando uma combinação de argumentos posicionais e nomeados
apresentar("Roberto", idade=27, cidade="Erechim")
```

9. valor padrão para desconto

```python
def preco_final(preco, desconto=0):
    return preco * (1 - desconto/100)

preco_final(50, 14)
```

10. função com vários parâmetros opcionais

```python
def produto(nome, preco, quantidade=1, desconto=0):
    return preco * quantidade * (1 - desconto/100)

produto("cachaça", 50, 5, 10)
```

11. modificação de um inteiro

```python
def modificar(x):
    x = x + 10

numero = 20
modificar(numero)
print(numero)
```
