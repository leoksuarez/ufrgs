1. função para calcular o dobro

```python
def dobro(n):
    return 2*n

dobro(7)
```

2. função para calcular a média

```python
def media(a, b, c):
    return (a+b+c)/3

media(3,2,7)
```

3. função para verificar paridade

```python
def eh_par(n):
    if n % 2 == 0:
        return "é par"
    else:
        return "não é par"
eh_par(7)
```

4. conversão de temperatura

```python
def celsius_para_fahrenheit(c):
    return (9/5)*c + 32

celsius_para_fahrenheit(30)
```

5. área de um retângulo

```python
def area_retangulo(base, altura):
    return base*altura

area_retangulo(4, 5)
```

7. potência com expoente padrão

```python
def potencia(base, expoente=2):
    return base**expoente

potencia(4, 3)
```

8. argumentos posicionais e nomeados

```python
def apresentar(nome, idade, cidade):
    print(f'Olá, meu nome é {nome}, tenho {idade} anos e moro em {cidade}.')

# a) usando apenas argumentos posicionais
apresentar('Leo', 23, 'Porto Alegre')

# b) usando apenas argumentos nomeados
apresentar(idade=79, cidade="São José dos Campos", nome="Laura")

# c) usando uma combinação de argumentos posicionais e nomeados
apresentar("Roberto", idade=27, cidade="Erechim")
```

9. valor padrão para desconto

```python
def preco_final(preco, desconto=0):
    return preco * (1 - desconto/100)

preco_final(50, 14)
```

10. função com vários parâmetros opcionais

```python
def produto(nome, preco, quantidade=1, desconto=0):
    return preco * quantidade * (1 - desconto/100)

produto("cachaça", 50, 5, 10)
```

11. modificação de um inteiro

```python
def modificar(x):
    x = x + 10

numero = 20
modificar(numero)
print(numero)
```

a variável `numero` não foi alterada porque recebeu um objeto do tipo `int`, que é imutável. ao fazer `x = x + 10`, um novo objeto é criado na memória para o qual a variável local `x` aponta. a variável de fora, `numero` não muda.

```python
def modificar_mutavel(x):
    return x + 10

numero = 20
numero = modificar_mutavel(numero)
print(numero)
```

agora, fazemos a função `modificar_mutavel` retornar um valor, que depois é atribuído à variável `numero`.

13. modificação de uma lista

```python
def adicionar(lista, valor):
    lista.append(valor)

numeros = [1, 2, 3]
adicionar(numeros, 4)
print(numeros)
```

a lista original foi modificada porque `list`é um objeto mutável, ou seja, quando executamos `lista.append`, a própria lista `numeros` é modificada.

14. lista e atribuição

```python
def modificar(lista):
    lista = lista + [10]

numeros = [1,2,3]
modificar(numeros)
print(numeros)
```

a diferença aqui é que o comando `lista + [10]` cria um novo objeto para o qual a variável local (dentro da função) `lista` aponta. no exemplo anterior, o método `append` modifica o próprio objeto (a lista original).

15. lista com append

```python
def duplicar_elementos(lista):
    for i in range(len(lista)):
        lista[i] *= 2

lista = [1, 2, 3]
duplicar_elementos(lista)
print(lista)
```

16. modificação de um dicionário

```python
def atualizar_aluno(aluno, idade):
    aluno['idade'] = idade

aluno = {"nome": "Ana", "idade": 20}
atualizar_aluno(aluno, 21)
print(aluno)
```

17. dicionário vs atribuição

```python
def modificar(pessoa):
    pessoa = {"nome": "Carlos"}

dados = {"nome": "Ana"}
modificar(dados)
print(dados)
```

o comando `{"nome": "Carlos"}` cria um novo objeto na memória, para o qual a variável local `pessoa` referencia. com isso, a variável de fora `dados` não é alterada.

18. função que modifica um conjunto

```python
def adicionar_elementos(conjunto, elementos):
    for elemento in elementos:
        conjunto.add(elemento)

numeros = {1, 2, 3}
adicionar_elementos(numeros, [3, 4, 5])
print(numeros)
```

19. função que retorna uma nova lista

```python
def pares(lista):
    return [x for x in lista if x % 2 == 0]

numeros = [1, 2, 3, 4, 5, 6]
resultado = pares(numeros)
print(resultado)
print(numeros)
```

20. desafio: cadastro de alunos
