4. calcular soma e média

```python
notas = [7.5, 8.0, 6.5, 9.0, 8.5]

soma = 0

for nota in notas:
    soma += nota
print(soma)
media = soma/len(notas)
print(media)
```

5. encontrar maior e menor elemento

```python
numeros = [45, 12, 78, 34, 9, 56]

maior = numeros[0]
menor = numeros[0]

for numero in numeros:
    if numero > maior:
        maior = numero
    elif numero < menor:
        menor = numero
print(f'maior = {maior}, menor = {menor}')
```

6. contar números pares

```python
inteiros = [2, 5, 52, 33, 90, 231]
pares = []

for inteiro in inteiros:
    if inteiro % 2 == 0:
        pares.append(inteiro)
print(pares)
print(f'existem {len(pares)} pares na lista')
```

7. remover elementos repetidos

8. inverter uma lista

```python
numeros = [1, 2, 3, 4, 5]
invertida = numeros[::-1]
print(invertida)
```

9. separar pares e ímpares

```python
numeros = [22, 4942, 329, 46, 67, 69, 80, 13, 65]
pares = []
impares = []

for numero in numeros:
    if numero % 2 == 0:
        pares.append(numero)
    else:
        impares.append(numero)
print(f'pares: {pares}\nimpares: {impares}')
```

10. ordenar uma lista

```python
numeros = [45, 12, 78, 3, 56, 21]
crescente = numeros.copy()
n = len(numeros)

for i in range(n):
    trocado = False
    for j in range(0, n-i-1):
        if crescente[j] > crescente[j+1]:
            crescente[j], crescente[j+1] = crescente[j+1], crescente[j]
            trocado = True
    if not trocado:
        break
print(f'ordem crescente: {crescente}\nordem decrescente: {crescente[::-1]}')
```

11. compreensão de listas

```python
quadrados = [i**2 for i in range(1, 21) if i % 2 == 0]
print(quadrados)
```

12. trabalhando com uma matriz

```python
matriz = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
soma = 0
n = len(matriz)
m = len(matriz[0])

for i in range(n):
    for j in range(m):
        soma += matriz[i][j]
print(f'soma dos elementos da matriz = {soma}')
```
