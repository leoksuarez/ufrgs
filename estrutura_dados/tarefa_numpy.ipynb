1. criando e analisando um array

```python
import numpy as np

x = np.array([2, 4, 6, 8, 10])

print(x.ndim)
print(x.size)
print(x.dtype)
```

2. operações vetoriais

```python
x = np.array([1, 2, 3, 4, 5])

print(2*x + 3)
```

3. criação de uma matriz

```python
A = np.arange(1, 17).reshape(4, 4)

print(A)
```

4. soma das linhas e colunas

```python
A = np.arange(1, 10).reshape(3, 3)

print(A)
n, m = np.shape(A)

for i in range(n):
   print(f'soma da linha {i} = {sum(A[i])}')
for j in range(m):
   print(f'soma da coluna {j} = {sum(A[:, j])}')
```

5. produto matricial

```python
A = np.array([
    [1, 2], [3, 4]
    ])

B = np.array([
    [5, 6],
    [7, 8]
])

print(A @ B)
print(np.matmul(A, B))
print(np.dot(A, B))
```

6. determinante e inversa

```python
import numpy.linalg as la

A = np.array([
    [4, 7],
    [2, 6]
])

print(f'det A = {la.det(A)}')
print(f'A inversa =\n{la.inv(A)}')
```

7. sistema linear

```python
A = np.array([
    [3, 2],
    [1, -1]
])

b = np.array([7, 1])

x = np.linalg.solve(A, b)
print(x)
```

8. autovalores

```python
A = np.array([
    [4, 1],
    [2, 3]
])

print(f'autovalores de A: {la.eigvals(A)}')
```

9. estatística

```python
notas = np.array([7, 8, 5, 9, 10, 6, 8, 7])

print(f'média: {np.mean(notas)}')
print(f'mediana: {np.median(notas)}')
print(f'variância: {np.var(notas)}')
print(f'desvio padrão: {np.std(notas)}')
```

10. seleção por condição

```python
x = np.array([3, 12, 7, 20, 5, 18, 2, 25])

print(x[x > 10])
```

11. aproximação de uma integral

```python
x = np.linspace(0, np.pi, 101)
y = np.sin(x)

I = np.trapezoid(y, x)

print(f'integral aproximada: {I:.8f}')
print(f'erro: {np.abs(I-2):.2e}')
```

12. aproximação da derivada

```python
x = np.linspace(0, 1, 10**6)

f = x**3
df = np.gradient(f, x)

erro = np.max(np.abs(df - 3*x**2))
print(f'erro máximo em [0,1]: {erro:.2e}')
```

13. broadcasting

```python
A = np.arange(1, 10).reshape(3, 3)
v = np.array([10, 20, 30])

print(A + v)
```

14. exercício adicional -- SVD

```python
A = np.arange(1, 7).reshape(3, 2)

U, S, Vt = np.linalg.svd(A, full_matrices=False)

print(f'U = {U}')
print(f'diag(S) = {np.diag(S)}')
print(f'Vt = {Vt}')

erro = la.norm(A - U @ np.diag(S) @ Vt)

print(f'erro: {erro:.2e}')
```

observa que a decomposição SVD possui erro $\mathcal{O}(10^{-15})$, que é pequeno, mas não é nulo.
