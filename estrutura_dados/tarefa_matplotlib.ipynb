1. gráfico de uma função

```python
import numpy as np 
import matplotlib.pyplot as plt

x = np.linspace(-5, 5)
plt.plot(x, x**2, label=r'$y=x^2$')
plt.xlabel(r'$x$')
plt.ylabel(r'$f(x)$')
plt.grid()
plt.legend()
plt.show()
```

2. duas funções

```python
x = np.linspace(-3, 8)

plt.plot(x, x**2, label=r'$f(x)=x^2$')
plt.plot(x, 2*x + 1, label=r'g(x)=2x+1')

plt.legend()
plt.grid()
plt.show()
```

3. gráfico de seno e cosseno

```python
x = np.linspace(0, 2*np.pi)

plt.plot(x, np.sin(x), label=r'$\sin(x)$')
plt.plot(x, np.cos(x), label=r'$\cos(x)$')

tick_pos = np.arange(0, 2.1 * np.pi, np.pi / 2)

tick_labels = [
    r'0', r'$\frac{\pi}{2}$', r'$\pi$', r'$\frac{3\pi}{2}$', r'$2\pi$'
]

plt.xticks(tick_pos, tick_labels)
plt.grid(True, linestyle='--')
plt.legend()
plt.show()
```

4. gráfico de barras

```python
quadro_notas = {"Ana": 8.5, "Bruno": 7.0, "Carlos": 9.0, "Daniel": 6.5, "Eduardo": 8.0}
alunos = list(quadro_notas.keys())
notas = list(quadro_notas.values())

plt.bar(alunos, notas)
plt.title("Notas dos alunos")
plt.show()
```

5. gráfico de dispersão

```python
horas = [h for h in range(1, 7)]
nota = [n for n in range(4, 10)]

plt.scatter(
    horas, nota,
    s=100,
    marker='o',
    alpha=0.7
)
plt.xlabel("Horas de estudo")
plt.ylabel("Nota")
plt.show()
```

6. histograma

```python
dados = np.random.normal(0, 1, 1000)

plt.hist(dados, bins=20)

plt.xlabel("Valor")
plt.ylabel("Frequência")
plt.title("Distribuição normal")

plt.show()
```

7. boxplot

11. gráfico de contorno

```python
x = np.linspace(-5, 5)
y = np.linspace(-5, 5)

X, Y = np.meshgrid(x, y)

Z = X**2 + Y**2

plt.contour(X, Y, Z)

plt.xlabel(r'$x$')
plt.ylabel(r'$y$')
plt.colorbar()

plt.grid()
plt.show()
```

12. campo vetorial

```python
x = np.linspace(-5, 5)
y = np.linspace(-5, 5)
X, Y = np.meshgrid(x, y)

R = np.sqrt(X**2 + Y**2)

U = np.cos(Y) / R
V = np.sin(Y) / R

fig = plt.figure()

campo = plt.quiver(X, Y, U, V, R, cmap="jet")
fig.colorbar(campo)

plt.xlim(-5, 5)
plt.ylim(-5, 5)
plt.xlabel(r'$x$')
plt.ylabel(r'$y$')
plt.grid()
plt.show()
```

13. gráfico polar

```python
theta = np.linspace(-2*np.pi, 2*np.pi, 200)
r = 4 * theta 

fig, ax = plt.subplots(subplot_kw={'projection': 'polar'})

ax.plot(theta, r)
plt.show()
```

14. superfície 3D

```python
x = np.linspace(-5, 5)
y = np.linspace(-5, 5)
X, Y = np.meshgrid(x, y)

Z = X**2 + Y**2

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

superficie = ax.plot_surface(X, Y, Z, cmap="viridis")
fig.colorbar(superficie)
plt.show()
```

15. comparação de quatro visualizações

```python
x = np.linspace(0, 2*np.pi)

# cria a figura
fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(2, 2, figsize=(10, 8))

# primeiro grafico: y=sin(x)+cos(x)
ax1.plot(x, np.sin(x) + np.cos(x))
ax1.set_title(r'$y=\sin(x)+\cos(x)$')

# segundo grafico: barra
times = ["Grêmio", "Internacional", "Juventude"]
torcedores = [10, 7.5, .03]
ax2.bar(times, torcedores)
ax2.set_title("Número de torcedores (em milhões)")

# terceiro grafico: histograma
dados = np.random.normal(0, 4, 1000)
ax3.hist(dados, bins=20)
ax3.set_title('Histograma')

# quarto grafico: dispersao
ax4.scatter([1, 2, 3, 4, 5, 6], [6, 4, 5, 1, 3, 4])
ax4.set_title('Dispersão')

# plota tudo
plt.tight_layout()
plt.show()
```

16. exercício adicional -- gráfico científico completo

```python
x = np.linspace(-10, 10, 300)
y = np.exp(-0.2*x) * np.sin(x)

plt.figure(figsize=(8, 5))
plt.plot(x, y, label=r'$f(x) = e^{-0.2x}\sin(x)$')
plt.axhline(y=0, color='black', linestyle='--', linewidth=0.8, label=r'$y=0$')

plt.fill_between(x, y, 0, where=(y > 0), color='green', alpha=0.3, label='f positiva')
plt.fill_between(x, y, 0, where=(y < 0), color='red', alpha=0.3, label='f negativa')

plt.title('gráfico do produto de funções')
plt.xlabel(r'$x$')
plt.ylabel(r'$f(x)$')
plt.grid(True)
plt.legend()
plt.show()
```
