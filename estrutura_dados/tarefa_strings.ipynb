1. inverter uma string

```python
texto = "eu gosto de andar de bicicleta"

texto_invertido = texto[::-1]
print(texto_invertido)
```

2. contar vogais

```python
def conta_vogais(texto):
    vogais = "aeiou"
    contagem = 0
    for caracter in texto.lower():
        if caracter in vogais:
            contagem += 1
    return contagem

texto = "Radiohead é a melhor banda do mundo"
print(conta_vogais(texto))
```

3. contar palavras

```python
def conta_palavras(frase):
    palavras = frase.split(" ")
    return len(palavras)

frase = "o grêmio será rebaixado"
print(conta_palavras(frase))    
```

5. verificar palíndromo

```python
def eh_palindromo(texto):
    if texto.lower() == texto.lower()[::-1]:
        return "é palíndromo"
    else:
        return "não é palíndromo"

texto = "radar"
print(f"'{texto}' {eh_palindromo(texto)}")
```

6. normalização de espaços

```python
frase = "a      vida  não tem  sentido"
frase_normalizada = " ".join(frase.split())

print(frase_normalizada)
```

13. frequência de caracteres

```python
def conta_caracter(texto):
    frequencias = {}
    for char in texto:
        if char in frequencias:
            frequencias[char] += 1
        else:
            frequencias[char] = 1
    return frequencias

texto = "estrutura de dados"
print(conta_caracter(texto))
```

14. separação de dados

```python
texto = "Marcelo;45;Porto Alegre;RS"

campos = texto.split(";")

for campo in campos:
    print(campo)
```
