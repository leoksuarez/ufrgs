1. usando `viridis`

```python
import numpy as np 
import matplotlib.pyplot as plt 

rng = np.random.default_rng()
n = 10
A = rng.random((n, n))

plt.imshow(A, cmap='viridis')
plt.colorbar()
plt.show()
```

2. comparando dois colormaps

```python
n = 5
A = rng.integers(low=0, high=100, size=(n, n))

fig, axes = plt.subplots(1, 2, figsize=(10, 4))

im1 = axes[0].imshow(A, cmap='plasma')
axes[0].set_title('plasma')
fig.colorbar(im1, ax=axes[0])

im2 = axes[1].imshow(A, cmap='inferno')
axes[1].set_title('inferno')
fig.colorbar(im2, ax=axes[1])

plt.tight_layout()
plt.show()
```

3. invertendo um colormap

```python
A = rng.integers(low=0, high=100, size=(n, n))

fig, axes = plt.subplots(1, 2, figsize=(10, 4))

im1 = axes[0].imshow(A, cmap='viridis')
axes[0].set_title('viridis')
fig.colorbar(im1, ax=axes[0])

im2 = axes[1].imshow(A, cmap='viridis_r')
axes[1].set_title('viridis_r')
fig.colorbar(im2, ax=axes[1])

plt.tight_layout()
plt.show()
```

4. colormap em gráfico de dispersão

```python
x = rng.random(200)
y = rng.random(200)
z = x**2 + y**2

plt.scatter(
    x, y,
    c=z,
    cmap='plasma',
    s=50
)

plt.colorbar(label=r'$z=x^2+y^2$')
plt.xlabel(r'$x$')
plt.ylabel(r'$y$')
plt.show()
```

5. valores positivos e negativos

```python
from matplotlib.colors import TwoSlopeNorm

norm = TwoSlopeNorm(
    vcenter=0,
    vmin=-10,
    vmax=10
)

A = rng.integers(low=-10, high=10, size=(5, 5))

plt.imshow(A, cmap='RdBu_r', norm=norm)

plt.colorbar()
plt.show()
```

6. criando um colormap personalizado

```python
from matplotlib.colors import LinearSegmentedColormap 

cores = ['blue', 'white', 'red']

meu_cmap = LinearSegmentedColormap.from_list('meu_mapa', cores)

plt.imshow(A, cmap=meu_cmap)
plt.colorbar()
plt.show()
```

7. colormap discreto

```python
from matplotlib.colors import ListedColormap

A = rng.integers(low=-10, high=10, size=(3, 3))

cores = [
    'blue',
    'green',
    'yellow',
    'orange',
    'red'
]

meu_cmap = ListedColormap(cores)

plt.imshow(A, cmap=meu_cmap)
plt.colorbar()
plt.show()
```

8. normalização logarítmica

```python
from matplotlib.colors import LogNorm

A = rng.integers(low=1, high=100000, size=(20, 20))

plt.imshow(
    A,
    cmap='magma',
    norm=LogNorm(vmin=1, vmax=100000)
)

plt.colorbar()
plt.show()
```

9. superfície 3D com colormap

```python
from mpl_toolkits.mplot3d import Axes3D

x = np.linspace(-5, 5, 200)
y = np.linspace(-5, 5, 200)

X, Y = np.meshgrid(x, y)
Z = np.sin(np.sqrt(X**2 + Y**2))

fig = plt.figure()
ax = fig.add_subplot(projection='3d')

surf = ax.plot_surface(
    X, Y, Z,
    cmap='viridis'
)

fig.colorbar(surf)
plt.show()
```

10. criando um mapa de erros

```python
norm = TwoSlopeNorm(
    vmin=-5,
    vcenter=0,
    vmax=5
)

erro = rng.normal(size=(20,20))

plt.imshow(
    erro,
    cmap='RdBu_r',
    norm=norm
)

plt.colorbar(label='Erro')
plt.title('Mapa de erros')
plt.show()
```
